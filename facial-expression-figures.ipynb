```python
from feat.plotting import plot_face
import numpy as np
```

```python
# AUs ordered as:
# 1, 2, 4, 5, 6, 7, 9, 11, 12, 14, 15, 17, 20, 23, 24, 25, 26, 28, 43
component1 = np.array([0.1109, 0.0857, 0.0997, 0.0698, 0, 0, 0.0178, 0.0636, 0.0046, 0.012, 0.0877, 0.1321, 0.0343, 0.045, 0.066, 0.1101, 0, 0, 0, 0])
component2 = np.array([0, 0, 0, 0, 0.1339, 0, 0.019, 0.2271, 0.2359, 0.1978, 0, 0, 0.0604, 0.0063, 0.0897, 0.0059, 0, 0, 0, 0])
component3 = np.array([0, 0, 0.1234, 0, 0.1279, 0.3464, 0.0792, 0, 0, 0.007, 0.0348, 0.0877, 0.0133, 0.0573, 0.0295, 0.0688, 0, 0, 0, 0])

muscles = {'all': 'heatmap'}

comp1 = plot_face(au=component1, title="Component 1")
comp1_heatmap = plot_face(au=component1, muscles=muscles, title="Component 1")

comp2 = plot_face(au=component2, title="Component 2")
comp2_heatmap = plot_face(au=component2, muscles=muscles, title="Component 2")

comp3 = plot_face(au=component3, title="Component 3")
comp3_heatmap = plot_face(au=component3, muscles=muscles, title="Component 3")
```
